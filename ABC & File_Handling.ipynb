# Abstract Method

```python
from abc import ABC, abstractmethod
```

```python
class car (ABC):
    @abstractmethod
    def airbag(self):
        pass
```

```python
class nexon(car):
    def airbag(self):
        return 'Airbags are installed in nexon'
    def button_start(self):
        return 'Car will Start'
```

```python
obj = nexon()
```

```python
obj.airbag()
```

```python
obj.button_start()
```

```python
class car (ABC):
    @abstractmethod
    def sunroof(self):
        pass
```

```python
class fortuner(car):
    def sunroof(self):
        return 'sunroof is opening'
```

```python
f1 = fortuner()
```

```python
f1.sunroof()
```

```python
class phone (ABC):
    @abstractmethod
    def call(self):
        pass
```

```python
class samsung(phone):
    def call(self):
        return 'Calling from samsung phone'
    def camera(self):
        return 'Opning camera from samsung mobile'
```

```python
f2 = samsung()
```

```python
f2.call()
```

```python
f2.camera()
```
