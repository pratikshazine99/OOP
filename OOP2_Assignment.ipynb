# Part A: Single Inheritance (Level 1)

```python
# 1.Create a parent class Animal with a method eat(). Create a child class Dog that inherits Animal and adds a method bark().
```

```python
class Animal():
    def eat(self):
        return 'Animal is eating'
```

```python
class Dog(Animal):
    def bark(self):
        return 'Dog is barking'
```

```python
obj1=Dog()
```

```python
obj1.eat()
```

```python
obj1.bark()
```

```python
# 2.Create a class Vehicle with brand and year. Create a child class Car that adds fuel_type.
```

```python
class Vehicle:
    def __init__(self,Brand,Year):
        self.Brand=Brand
        self.Year=Year        
```

```python
class Car(Vehicle):
    def __init__(self,Brand,Year,Fuel_type):
        self.Brand=Brand
        self.Year=Year
        self.Fuel_type=Fuel_type
    def start(self):
        print(self,Brand,Year,Fuel_type)
```

```python
obj2=Car('Nexon',2023,'Petrol')
```

```python
obj2.Brand
```

```python
obj2.Year
```

```python
obj2.Fuel_type
```

```python
# 3.Write a class Person with name. Create a child Employee with employee_id.
```

```python
class Person:
    def __init__(self,Name):
        self.Name=Name
    
```

```python
class Employee(Person):
    def __init__(self,Name,Employee_id):
        self.Name=Name
        self.Employee_id=Employee_id
    def info(self):
        print(self,Name,Employee_id)
```

```python
obj3=Employee('Snehal',2421)
```

```python
obj3.Name
```

```python
obj3.Employee_id
```

## Part B: Multilevel Inheritance (Level 2)

```python
# 1.The Tech Chain: Create a class ElectronicDevice. Inherit it into Computer. Then inherit Computer into Laptop. Add one unique method to each.
```

```python
class ElectronicDevice():
    def Keyboard(self):
        return 'KeyBoard Brand name is Dell'
```

```python
class Computer(ElectronicDevice):
    def brand(self):
        return 'Computer Brand Name is Dell'
```

```python
class Laptop(Computer):
    def Ram(self):
        return 'Laptop has 16GB Ram'
```

```python
obj=Laptop()
```

```python
obj.Keyboard()
```

```python
obj.brand()
```

```python
obj.Ram()
```

```python
# 2.Create a Camera class and a Phone class. Create a Smartphone class that inherits from both and has a method take_selfie().
```

```python
class Camera():
    def click_photo(self):
        return 'Camera can click photo'
```

```python
class Phone():
    def call(self):
        return 'Phone can make calls'
    
```

```python
class Smartphone(Camera,Phone):
    def take_selfie(self):
        return 'Smartphone can take selfie'
```

```python
obj=Smartphone()
```

```python
obj.take_selfie()
```

```python
obj.call()
```

```python
obj.click_photo()
```
