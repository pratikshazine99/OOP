```python
class bank():
    def __init__(self,balance):
        self.balance = balance
    def intrest(self):
        return self.balance*0.10
    def fd(self,months):
            total=self.balance*(0.10*months)
            return total
```

```python
cust1=bank(5000)
```

```python
cust1.balance
```

```python
cust1.intrest()
```

```python
cust1.fd(20)
```

# Incapsulation

# Inheritance


```python
class parent():
    def singing(self):
        return 'I am a parent class and having skill of singing'
```

```python
class child(parent):
    def cricket(self):
        return 'I am a child class and having skills of playing cricket'
```

```python
obj2=child()
```

```python
obj2.cricket()
```

```python
obj2.singing()
```

```python
obj3=parent()
```

```python
obj3.singing()
```

```python
class car():
    def __init__(self,engine,fuel,seat,color): # constructor way to pass data to class
        self.engine = engine
        self.fuel = fuel
        self.seat = seat
        self.color = color
    def start(self):
        return 'Car is starting'
```

```python
class bus(car):
    def off(self):
        return 'Bus is getting off'
```

```python
obj4 = bus(500,'Diesel',30,'White')
```

```python
obj4.start()
```

```python
obj4.off()
```

```python
obj5 = car(500,'Diesel',30,'White')
```

```python
obj5.start()
```

```python
class bike(bus):
    def kickstart(self):
        return 'Bike will start'
```

```python
obj6 = bike(500,'Diesel',30,'White')
```

```python
obj6.kickstart()
```
